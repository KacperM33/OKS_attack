## Konwersja sekwencji wideo na filmy w formacie mkv

```python
import os
import subprocess
from tqdm import tqdm # pasek postępu

# ścieżka do sekwencji klatek
frames_folder = "../penn-action-dataset/frames"         # <--- ścieżka do folderu z folderami z sekwencjami
# folder wyjściowy (do zapisu wyników)
output_folder = "../penn-action-dataset/videos"
# lista plików do przetworzenia
files_list = "../penn-action-dataset/all_test_subset.txt"

# tworzenie folderu wyjściowego jeśli nie istnieje
os.makedirs(output_folder, exist_ok=True)

# filtrowanie aby mieć katalogi z klatkami
folders = []
with open(files_list, 'r') as f:
    for line in f:
        folder_name = line.strip()
        if folder_name:
            folders.append(folder_name)

folders = sorted(folders)

print(f"Znaleziono {len(folders)} wideo do przetworzenia z pliku .txt...\n")

errors_counter = 0

# pętla przetwarzająca
for folder_name in tqdm(folders, desc="Konwertowanie"):
    
    input_folder_path = os.path.join(frames_folder, folder_name)
    if not os.path.exists(input_folder_path):
        tqdm.write(f"UWAGA: Folder {input_folder_path} nie istnieje. Pomijam...")
        continue

    output_video_path = os.path.join(output_folder, f"{folder_name}-vis.mkv")

    # Wzorzec: "%06d.png" (%06d oznacza 6 cyfr: 000001, 000002...)
    file_pattern = f'%06d.jpg' 

    input_pattern_path = os.path.join(input_folder_path, file_pattern)

    print(input_pattern_path)

    # FFMPEG
    # cmd = (
    #     f'ffmpeg -y -framerate 25 '
    #     f'-hide_banner -loglevel error '
    #     f'-i "{input_pattern_path}" '
    #     f'-c:v libx264 -pix_fmt yuv420p '
    #     f'"{output_video_path}"'
    # )


    # ffmpeg -y -framerate 30 -i penn-action-dataset/frames/0119/%06d.jpg -c:v ffv1 -level 3 -coder 1 -context 1 -slicecrc 1 penn-action-dataset/videos/0119.mkv
    cmd = (
        f'ffmpeg -y -framerate 30 '
        f'-i "{input_pattern_path}" '
        f'-c:v ffv1 -level 3 -coder 1 -context 1 -slicecrc 1 '
        f'"{output_video_path}"'
    )

    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)

    if result.returncode != 0:
        print(f"\n[!] BŁĄD przy folderze {folder_name}.")
        print("--- LOGI FFMPEG ---")
        print(result.stderr) # FFmpeg wyrzuca wszystko do stderr
        print("-------------------")
        
        errors_counter += 1
        if os.path.exists(output_video_path):
            os.remove(output_video_path)


print("\nZakończono konwersję sekwencji klatek do wideo.")
print(f"Poprawnie przekonwertowano {len(folders) - errors_counter}/{len(folders)} plików.")
print(f"Nie udało się przekonwertować: {errors_counter} plików.")
```

Sprawdzenie jakich wideo brakuje

```python
import os

output_folder = "../penn-action-dataset/videos"
files_list = "../penn-action-dataset/all_test_subset.txt"

# Pobieramy listę oczekiwanych folderów
expected_folders = set()
with open(files_list, 'r') as f:
    for line in f:
        folder_name = line.strip()
        if folder_name:
            expected_folders.add(folder_name)

# Sprawdzamy co faktycznie jest na dysku
generated_videos = set()
if os.path.exists(output_folder):
    for file in os.listdir(output_folder):
        if file.endswith("-vis.mkv"):
            # Ucinamy "-vis.mkv", żeby mieć samą nazwę folderu
            folder_name = file.replace("-vis.mkv", "")
            generated_videos.add(folder_name)

# Różnica zbiorów pokaże nam brakujące pliki
missing = expected_folders - generated_videos

print(f"Oczekiwano: {len(expected_folders)}, Wygenerowano: {len(generated_videos)}")
print(f"Brakujące/Błędne ({len(missing)}):")
for m in sorted(missing):
    print(f"- {m}")
```
